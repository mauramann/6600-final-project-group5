# Clean Raw Data

```python
import pandas as pd

# Load Data Set
data = pd.read_excel("data/raw_data.xlsx")
data.head
```

```python
# Search for NAs
data.isna().sum()
```

```python
# Evaluate NA Rows - Fine to keep because they don't affect the Description Column
na_rows = data[data.isna().any(axis=1)]
na_rows
```

```python
# Check data types
data.dtypes
```

```python
# Address outliers

# Find "Testing" rows - not relevant observations
test_rows = data[data["Subject"].str.contains("test", na=False)]

# Drop testing rows
new_data = data.drop(test_rows.index)
```

```python
# Drop Unwanted Columns
final_data = new_data.drop(columns=['Subject', 'Created On'])
```

```python
# Print old shape
print(data.shape)
# Print new data shape
print(final_data.shape)
```

```python
# Write out clean data
final_data.to_csv("data/clean_data.csv")
```
